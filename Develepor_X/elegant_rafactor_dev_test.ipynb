## Rock Drawing Refactor Flow Test

This section tests the new `RockRenderContext` + `DrawMachine` drawing flow without touching breeding/game-state code.


```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == 'Develepor_X':
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import matplotlib.pyplot as plt
from importlib import reload

import Rock_Genetics.rock_genetic_helper as genetics
import Rock_Drawing.rock_drawing_helper as drawing

genetics = reload(genetics)
drawing = reload(drawing)

print('loaded', drawing.RockRenderContext.__name__, drawing.DrawMachine.__name__)
```

```python
def make_forced_rock(rock_id=100, sex=genetics.Sex.MALE, choices=None, given='Refactor'):
    choices = choices or {}
    genes = {}

    for gene_name, spec in genetics.GENE_SPECS.items():
        allele_value = choices.get(gene_name, min(spec.options.keys()))
        genes[gene_name] = genetics.GenePair(
            allele_a=genetics.Allele(allele_value),
            allele_b=genetics.Allele(allele_value),
            name_of_gene=gene_name,
            dominance_type=spec.expression_rule,
        )

    rock = genetics.Rock(
        id=rock_id,
        sex=sex,
        name=genetics.RockName(given=given),
        genotype=genetics.Genome(genes),
    )

    genetics.ExpressionEngine().instantiate_phenotype(rock)
    return rock

def make_random_rock(rock_id=101, sex=genetics.Sex.MALE, given='Random'):
    genome = genetics.GenomeFactory().make_random_rock_genome()
    rock = genetics.Rock(
        id=rock_id,
        sex=sex,
        name=genetics.RockName(given=given),
        genotype=genome,
    )
    genetics.ExpressionEngine().instantiate_phenotype(rock)
    return rock
```

```python
feature_choices = {
    'shape': 3,
    'size': 0,
    'color': 6,
    'eyes': 1,
    'brows': 3,
    'mouths': 4,
    'noses': 4,
    'arms': 2,
    'crowns': 3,
    'wings': 1,
    'halos': 1,
    'horns': 1,
    'wrinkles': 1,
    'fuzz': 1,
    'hair': 1,
    'facial_hair': 6,
    'freckles': 1,
    'stones': 1,
    'tails': 1,
    'eye_color': 6,
    'hair_color': 5,
    'ears': 4,
    'hair_texture': 1,
    'splitting': 0,
}

rocks = [
    make_random_rock(rock_id=201, given='Random'),
    make_forced_rock(rock_id=202, choices=feature_choices, given='Feature Stack'),
]

fig, axes = plt.subplots(1, len(rocks), figsize=(5 * len(rocks), 5))
if len(rocks) == 1:
    axes = [axes]

for ax, rock in zip(axes, rocks):
    machine = drawing.DrawMachine(rock=rock, ax=ax, normalize_size=True)
    machine.draw()
    print('rock', rock.id, drawing.get_visual_phenotype(rock))
    print('face slots', machine.ctx.face_layout.slots)

plt.tight_layout()
plt.show()
```
