```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

print(PROJECT_ROOT)

import Rock_Genetics.rock_genetic_helper as genetics
import Rock_Breeding.rock_breeding_helper as breeding


```

```python
factorio = genetics.GenomeFactory()
espresso = genetics.ExpressionEngine()
valuo = genetics.ValueCalculator()

breedo = breeding.BreedingMaster(
    GenomeFactory = factorio,
    ExpressionEngine = espresso,
    ValueCalculator = valuo,
)

Rock1 = genetics.Rock(
    id=1, 
    name="bruh", 
    sex=genetics.Sex.MALE, 
    genotype = factorio.make_random_rock_genome(),
    death_genes = factorio.make_death_genes(),
    generation = 0
    )

Rock2 = genetics.Rock(
    id=2, 
    name="bruha", 
    sex=genetics.Sex.FEMALE, 
    genotype = factorio.make_random_rock_genome(),
    death_genes =  factorio.make_death_genes(),
    generation = 0
)

parent_clutch = breedo.breed_parent_set(
    parent_a = Rock1,
    parent_b = Rock2,
    next_id = 3,
    child_generation = 1,
)

print(len(parent_clutch))

for child in parent_clutch:
    print(f"{child.id}")
    print(f"{child.genotype.print_genes()}")
    print(f"{child.death_genes.print_genes()}")
```
